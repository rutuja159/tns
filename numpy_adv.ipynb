```python
import numpy as np
```

```python
# ---- THE DATA ----

student_names = np.array(["Aisha", "Ben", "Carla", "Dev", "Eli", "Farah"])   # six students
subject_names = np.array(["Maths", "Physics", "English"])                    # three subjects

scores = np.array([[78, 65, 82],    # Aisha's three marks - one per subject, in the order above
                   [55, 71, 60],    # Ben's
                   [91, 88, 79],    # Carla's
                   [42, 38, 55],    # Dev's
                   [67, 74, 71],    # Eli's
                   [83, 97, 68]])   # Farah's

print(scores)                       # the whole table
print("shape:", scores.shape)
```

```python
total_marks=np.sum(scores,axis=1)
for name , total_marks in zip(student_names, total_marks):
    print(name, ":", total_marks)
```

```python
avg=np.mean(scores,axis=1)
for name , avg in zip(student_names, avg):
    print(name, ":", avg)
```

```python
average_marks = np.mean(scores,axis=0)
print(average_marks)
```

```python
print("Hard subject is",subject_names[subject_names.argmin()])
print("Easy subject is",subject_names[subject_names.argmax()])
```
